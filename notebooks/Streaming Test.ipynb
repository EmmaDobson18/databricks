```python
from pyspark.sql.functions import *
```

```python
%sql
USE workspace.default;

CREATE VOLUME IF NOT EXISTS streaming_checkpoints;

DROP TABLE IF EXISTS test_stream_emma;

CREATE TABLE IF NOT EXISTS test_stream_emma;
```

```python
checkpoint = "/Volumes/workspace/default/streaming_checkpoints/test_stream_4"
```

```python
df = (
    spark.readStream
         .format("rate")
         .option("rowsPerSecond", 1)
         .load()
)
```

```python
df.writeStream \
    .format("delta") \
    .trigger(availableNow=True) \
    .option("checkpointLocation", checkpoint) \
    .option("mergeSchema", "true") \
    .toTable("workspace.default.test_stream_emma")
```

```python
%sql
SELECT count(*) FROM test_stream_emma
LIMIT 100
```
